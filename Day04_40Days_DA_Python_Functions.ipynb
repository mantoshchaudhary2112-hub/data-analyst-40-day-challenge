### Return Values

```python
print() displays a result on the screen.
return sends a result back to the code that called the function, so it can be stored or used in further calculations.
```

```python
def calculate_revenue(price, quantity):
    revenue = price * quantity
    return revenue

total_revenue = calculate_revenue(750, 8)

print("Total revenue:", total_revenue)
print("Revenue after ₹500 discount:", total_revenue - 500)
```

### Mini-project: Sales Analysis Calculator

```python
def avg(total_sales, no_of_sales):
    return total_sales / no_of_sales

Sales = input("Enter sales separated by commas: ")

Sales = [float(sale.strip()) for sale in Sales.split(",")]

if len(Sales) > 0:
    no_of_sales = len(Sales)
    Total_Sales = sum(Sales)
    Average_Sales = avg(Total_Sales, no_of_sales)

    print("Number of sales:", no_of_sales)
    print("Total sales:", Total_Sales)
    print("Average sales:", Average_Sales)
else:
    print("Please enter at least one sale.")

def Total_Sales(Sales):
    Sales = input("Enter sales separated by commas: ")
    Sales = [float(sale.strip()) for sale in Sales.split(",")]
    Total_Sales = sum(Sales)
    return Total_Sales

def high(Sales):
    Sales = input("Enter sales separated by commas: ")
    Sales = [float(sale.strip()) for sale in Sales.split(",")]
    Max = max(Sales)
    return Max

def min(Sales):
    Sales = input("Enter sales separated by commas: ")
    Sales = [float(sale.strip()) for sale in Sales.split(",")]
    Min = min(Sales)
    return Min
```

```python
def calculate_average(total_sales, no_of_sales):
    return total_sales / no_of_sales

def calculate_total(sales):
    return sum(sales)

def find_highest(sales):
    return max(sales)

def find_lowest(sales):
    return min(sales)


try:
    Sales_Input = input("Enter sales separated by commas: ")

    if not Sales_Input.strip():
        print("Error: Please enter at least one sale.")

    elif any(value.strip() == "" for value in Sales_Input.split(",")):
        print("Error: Remove empty values from your input.")

    else:
        Sales = [float(value.strip()) for value in Sales_Input.split(",")]

        if any(sale < 0 for sale in Sales):
            print("Error: Sales cannot be negative.")

        else:
            no_of_sales = len(Sales)
            Total = calculate_total(Sales)
            Average = calculate_average(Total, no_of_sales)
            Highest = find_highest(Sales)
            Lowest = find_lowest(Sales)

            print("\n--- Sales Analysis Report ---")
            print("Number of sales:", no_of_sales)
            print("Total sales:", Total)
            print("Average sales:", Average)
            print("Highest sale:", Highest)
            print("Lowest sale:", Lowest)

except ValueError:
    print("Error: Enter numbers only, separated by commas.")
```
