```python

import pandas as pd

sales_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop"],
    "Category": ["Electronics", "Accessories", "Accessories", "Electronics", "Electronics"],
    "Quantity": [2, 10, 5, 3, 1],
    "Unit_Price": [50000, 500, 1500, 12000, 50000]
}

df = pd.DataFrame(sales_data)

df
```

### sales analysis

```python
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

df
```

```python
1. What is Pandas?

Interview-ready answer: Pandas is a Python library used to load, clean, manipulate, analyze, and organize structured data.




2. What is a DataFrame?

Interview-ready answer: A DataFrame is a two-dimensional tabular data structure with labeled rows and columns. 
Different columns can have different data types.




3. Difference between head() and info()

- df.head() displays the first 5 rows by default. You can change the number, e.g. df.head(3).
- df.info() provides a summary including column names, non-null counts, data types, and memory usage.



4. What does df.shape return?

Small correction: Use df.shape, not df.shape().

shape is an attribute, not a method. It returns a tuple in the format (rows, columns).

Example: (5, 4) means 5 rows and 4 columns.



5. How do you check missing values?

df.isnull().sum()



- isnull() identifies missing values.
- sum() counts them in each column.
```

```python
# Calculate total revenue
print("Total Revenue:", df["Total_Sales"].sum())
```

```python
# Find the best-selling product by revenue
df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
```

```python
# Find the highest-revenue category
category_sales = df.groupby("Category")["Total_Sales"].sum()

print(category_sales.idxmax())
print(category_sales.max())
```
