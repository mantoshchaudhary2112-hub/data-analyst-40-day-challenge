## String

```python
What is a string?
A string is a sequence of characters used to store text. In Python, strings are written inside single or double quotation marks.
```

```python
customer_name = "Mantosh"
city = "Jaipur"
product = 'Laptop'

print(customer_name)
print(city)
print(product)
```

```python
customer_name = "Mantosh"
product_name = "Laptop"
city = "Jaipur"

print("Customer:", customer_name)
print("Product:", product_name)
print("City:", city)
print(type(customer_name))
```

## String indexing and slicing

### Indexing

```python
Each character has a position, starting from index 0 
```

```python
word = "PYTHON"

print(word)
print(word[0])
print(word[2])
print(word[-1])
```

### Slicing

```python
word = "PYTHON"

print(word[0:3])
print(word[2:5])
print(word[:4])
print(word[3:])
```

```python
product_code = "LAPTOP2026"

print(product_code[0])
print(product_code[-1])
print(product_code[0:6])
print(product_code[6:])
```

### String methods

```python
.strip()
.lower()
.upper()
.title()
.replace()
.split()
```

```python
# Basic string-cleaning methods
customer = "  rAHUL shARMA  "

print("Cleaned:", customer.strip().title())
print("Lowercase:", customer.lower())
print("Uppercase:", customer.upper())
```

```python
# replace() and split()
sales_input = "Laptop, Mouse, Keyboard, Monitor"

cleaned_input = sales_input.replace("Monitor", "Printer")
products = cleaned_input.split(",")

print("Updated text:", cleaned_input)
print("Product list:", products)
print("Number of products:", len(products))
```

```python
# Clean spaces after splitting
customer_names = "  Rahul Sharma, PRIYA SINGH,  amit kumar  "

names = [
    name.strip().title()
    for name in customer_names.split(",")
]

print(names)
```

### Python file handling

```python
# Write a data to a file 
with open("sales_report.txt", "w") as file:
    file.write("Sales Report\n")
    file.write("Total Sales: 25000\n")
    file.write("Total Orders: 50\n")

print("Report saved successfully.")
```

```python
# Read data from a file 
with open("sales_report.txt", "r") as file:
    content = file.read()

print(content)
```

```python
with open("customer_report.txt", "w") as file:
    file.write("Customer Report\n")
    file.write("Customer: Mantosh\n")
    file.write("City: Jaipur\n")

with open("customer_report.txt", "r") as file:
    report = file.read()

print(report)
```

```python
# Corrected code 

customer_names = "  rAHUL shARMA, PRIYA SINGH ,  amit kumar, Neha Gupta  "

# Split and clean the names
cleaned_names = [
    name.strip().title()
    for name in customer_names.split(",")
]

print(cleaned_names)

# Count the customers

print("Total customers:", len(cleaned_names))

# save the report 
with open("cleaned_customer_report.txt", "w") as file:
    file.write("Customer Data Cleaning Report\n")
    file.write("============================\n")

    for name in cleaned_names:
        file.write(name + "\n")

    file.write(f"\nTotal customers: {len(cleaned_names)}")

print("Report saved successfully!")
```
