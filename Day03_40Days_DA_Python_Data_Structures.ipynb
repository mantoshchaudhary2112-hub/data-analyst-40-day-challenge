### Lists
A Python list stores multiple items in a single variable.
#### Mutable 

```python
# Day 3 - Task 1: Python Lists

Daily_Sales = [1200, 1500, 1800, 1300, 2200]

print("Daily Sales:", Daily_Sales)
print("First Day Sales:", Daily_Sales[0])
print("Third Day Sales:", Daily_Sales[2])
print("Total Sales:", sum(Daily_Sales))
print("Average Sales:", sum(Daily_Sales) / len(Daily_Sales))
```

```python
Products = ["Laptop", "Mouse", "Keyboard"]

# Add a product
Products.append("Monitor")

# Update the second product
Products[1] = "Wireless Mouse"

# Remove the third product
Products.remove("Keyboard")

print("Final Product List:", Products)
print("Number of Products:", len(Products))
```

```python
Sales = [2000, 3500, 1500, 4000]

Sales.append(3000)
Sales[0] = 2500
Sales.remove(1500)

print("Updated Sales:", Sales)
print("Total Sales:", sum(Sales))
print("Highest Sale:", max(Sales))
```

### Tuple
A tuple stores multiple items in an ordered collection, but unlike a list, its elements cannot be reassigned after creation.
#### Immutable 

```python
Employee = ("Mantosh", "Data Analyst Intern", 25000)

print("Name:", Employee[0])
print("Role:", Employee[1])
print("Monthly Salary:", Employee[2])
```

```python
Product = ("Laptop", 55000, "Electronics")

print("Product Name:", Product[0])
print("Price:", Product[1])

print("Product Details:", Product)
```

```python
Product = ("Laptop", 55000, "Electronics")

Product[1] = 60000  # TypeError
```

### Dictionaries
A dictionary stores data as key-value pairs.
For example, a product has a name, price, and category. Each key identifies its corresponding value.
#### Mutable 

```python
Product = {
    "Name": "Laptop",
    "Price": 55000,
    "Category": "Electronics"
}

print("Product Name:", Product["Name"])
print("Original Price:", Product["Price"])

# Update the price
Product["Price"] = 60000

# Add a new key-value pair
Product["Brand"] = "HP"

print("Updated Product:", Product)
```

```python
# Example 
Sales_Report = {
    "Product": "Laptop",
    "Units_Sold": 10,
    "Unit_Price": 50000
}

# Add total revenue
Sales_Report["Total_Revenue"] = (
    Sales_Report["Units_Sold"] * Sales_Report["Unit_Price"]
)

# Update units sold
Sales_Report["Units_Sold"] = 12

print(Sales_Report)
```

```python
Sales_Report["Units_Sold"] = 12
print(Sales_Report) 
'''
The issue is that Python calculated the revenue once and stored the result as a number. 
It doesn't automatically recalculate that number when Units_Sold changes.'''
```

```python
Sales_Report = {
    "Units_Sold": 10,
    "Unit_Price": 50000
}

Sales_Report["Total_Revenue"] = (
    Sales_Report["Units_Sold"] * Sales_Report["Unit_Price"]
)

Sales_Report["Units_Sold"] = 12

print(Sales_Report["Total_Revenue"])  # 500000
```

```python
# Updated Sales_Report
Sales_Report["Total_Revenue"] = (
    Sales_Report["Units_Sold"] * Sales_Report["Unit_Price"]
)

print(Sales_Report["Total_Revenue"])  # 600000
```

```python
# Day 3 - Product Sales Analyzer

Products = ["Laptop", "Mouse", "Keyboard"]

Store_Info = ("Mantosh Store", "Jaipur")

Sales_Data = {
    "Laptop": 2,
    "Mouse": 10,
    "Keyboard": 5
}

Prices = {
    "Laptop": 50000,
    "Mouse": 500,
    "Keyboard": 1000
}

print("Store:", Store_Info[0])
print("Location:", Store_Info[1])
print("Products:", Products)

Total_Revenue = 0

for Product in Products:
    Revenue = Sales_Data[Product] * Prices[Product]
    Total_Revenue += Revenue
    print(Product, "| Revenue:", Revenue)

print("Total Store Revenue:", Total_Revenue)
```
