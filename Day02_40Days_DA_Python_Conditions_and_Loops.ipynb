### Student Result Checker

```python
A student's marks determine their result:
- Marks of 80 or above: Excellent
- Marks of 60–79: First Division
- Marks of 40–59: Pass
- Marks below 40: Fail
```

```python
Marks = float(input("Enter total marks obtained "))

if (Marks>=80):
    print("Excellent")
elif (Marks>=60):
    print("First Division")
elif (Marks>=40):
    print("Pass")
else:
    print("Fail")
```

### Sales Performance Checker

```python
A sales employee's monthly sales determine their performance:
- Sales of ₹100,000 or more → Excellent Performance
- Sales of ₹70,000 or more → Good Performance
- Sales of ₹40,000 or more → Average Performance
- Sales below ₹40,000 → Needs Improvement
```

```python
Sales = float(input("Enter Your Monthly sales "))

if Sales < 0:
    print("Invalid sales amount")
elif (Sales>=100000):
    print("Excellent Performance")
elif (Sales>=70000):
    print("Good Performance")
elif (Sales>=40000):
    print("Average Performance")
else:
    print("Needs Improvement")
```

### Sales Target Tracker

```python
Problem: A shop has a monthly sales target of ₹50,000. Write a Python program that:
1. Uses a for loop to process sales for 5 days.
2. Asks the user to enter sales for each day.
3. Adds all daily sales to calculate total sales.
4. Uses if/else to check whether total sales reached ₹50,000.
```

```python
N = int(input("Number of days to calculate : "))
if N < 1 or N > 30:
    print("Invalid input! Enter days between 1 and 30.")

else:
    Total_Sales=0
    for day in range(1,N+1):
        Sales= int(input("Enter total sales "))
        Total_Sales += Sales
        print("Day",day,"   " , "Sales ",Sales)
    print("Total Sales : " ,Total_Sales) 
    if (Total_Sales>=50000):
        print("Sales Target achieved ")
    else:
        print("Need improvement Sales target not Reached ")
```
