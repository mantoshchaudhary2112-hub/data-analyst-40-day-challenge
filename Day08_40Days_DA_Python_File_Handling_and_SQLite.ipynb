```python
import csv

with open("sales.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["Product", "Sales"])
    writer.writerow(["Laptop", 50000])
    writer.writerow(["Mouse", 5000])
    writer.writerow(["Keyboard", 10000])
    writer.writerow(["Monitor", 15000])

with open("sales.csv", "r") as file:
    print(file.read())
```

```python
import csv

with open("sales.csv", "r", newline="") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
# Calculate total sales from CSV

import csv

total_sales = 0

with open("sales.csv", "r", newline="") as file:
    reader = csv.reader(file)

    next(reader)  # Skip the header row

    for row in reader:
        sales = int(row[1])
        total_sales += sales

print("Total Sales:", total_sales)
```

```python
# Find the highest-selling product

import csv

max_sales = 0
product_name = ""

with open("sales.csv", "r", newline="") as file:
    reader = csv.reader(file)

    next(reader)  # Skip header

    for row in reader:
        product = row[0]
        sales = int(row[1])

        if sales > max_sales:
            max_sales = sales
            product_name = product

print("Highest Selling Product:", product_name)
print("Highest Sales:", max_sales)
```

```python
# Create a Sales table

import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS Sales (
    Product TEXT,
    Quantity INTEGER,
    Price INTEGER
)
""")

connection.commit()

print("Sales table created successfully!")
```

```python
# Insert records into SQLite

import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute("""
INSERT INTO Sales (Product, Quantity, Price)
VALUES ('Laptop', 5, 50000)
""")

cursor.execute("""
INSERT INTO Sales (Product, Quantity, Price)
VALUES ('Mouse', 10, 500)
""")

cursor.execute("""
INSERT INTO Sales (Product, Quantity, Price)
VALUES ('Keyboard', 8, 1000)
""")

connection.commit()

print("Records inserted successfully!")
```

```python
# Retrieve records using SELECT

import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM Sales")

records = cursor.fetchall()

for row in records:
    print(row)

connection.close()
```

```python
# Calculate total sales value
# Total Value = Quantity x Prize 

import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute("""
SELECT SUM(Quantity * Price)
FROM Sales
""")

result = cursor.fetchone()

print("Total Sales Value:", result[0])

connection.close()
```

```python
# Sort records using ORDER BY

import sqlite3

connection = sqlite3.connect("sales.db")
cursor = connection.cursor()

cursor.execute("""
SELECT * FROM Sales
ORDER BY Price DESC
""")

records = cursor.fetchall()

for row in records:
    print(row)

connection.close()
```
