### Python modules
A module is a Python file or library containing reusable code, such as functions and variables, that we can import into another program.

```python
import math

number = 144

print("Square root:", math.sqrt(number))
print("Rounded up:", math.ceil(4.3))
print("Rounded down:", math.floor(4.8))
```

### Your first try and except

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number
    print("Result:", result)

except ZeroDivisionError:
    print("Error: You cannot divide by zero.")

except ValueError:
    print("Error: Please enter a valid number.")
```

### Sales Calculator mini-project

```python
Create and  write the program yourself using these requirements:
1. Ask the user to enter total sales revenue.
2. Ask the user to enter total expenses.
3. Convert both inputs to numbers using float().
4. Calculate profit using profit = revenue - expenses.
5. Print the revenue, expenses, and profit.
6. Use try and except ValueError to handle non-numeric input.
```

```python
try :
    Sales_Revenue = float(input("Enter Total Sales Revenue : "))
    Total_Expenses = float(input("Enter Total Total Expenses : "))
    Profit = Sales_Revenue - Total_Expenses
    print("Sales Revenue ; ",Sales_Revenue)
    print("Total Expenses : ",Total_Expenses)
    print("Profit : ",Profit)
except:
    print("Error: Please enter valid numbers.")

```
