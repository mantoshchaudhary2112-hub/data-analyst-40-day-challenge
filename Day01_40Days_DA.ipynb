### Calculate profit-Loss or Break-even

```python
quantity = 25
cost_per_item = 75
selling_price_per_item = 120

total_revenue = quantity * selling_price_per_item
total_cost = quantity * cost_per_item
profit = total_revenue - total_cost

print(f"Total revenue: ₹{total_revenue:.2f}")
print(f"Total cost: ₹{total_cost:.2f}")
print(f"Profit / loss: ₹{profit:.2f}")

if profit > 0:
    print("Business made a profit.")
elif profit < 0:
    print("Business made a loss.")
else:
    print("Business broke even.")
```

### Stores Sales and advertising

```python
'''Given information
- Quantity sold = 40
- Selling price per item = ₹650
- Cost price per item = ₹420
- Advertising expense = ₹1,500

### Formulas
1. **Total revenue** = quantity × selling price per item
2. **Total product cost** = quantity × cost price per item
3. **Profit before advertising** = total revenue − total product cost
4. **Final profit** = profit before advertising − advertising expense
5. **Profit margin (%)** = (final profit ÷ total revenue) × 100
6. **Business status** = profit if final profit > 0; loss if < 0; break-even if = 0
'''

Quantity = int(input(" Enter total Quantity Sold : "))
Sell_prize = int(input(" Enter total Selling prize per item  : "))
Cost_prize = int(input(" Enter total cost prize  : "))
Add_exp = int(input(" Enter Advertising expense  : "))
Profit = Sell_prize - Cost_prize

Total_revenue = Quantity * Sell_prize
Total_product_cost = Quantity * Cost_prize
Profit_before_advertising = Total_revenue - Cost_prize
Final_profit = Profit_before_advertising - Add_exp
Profit_margin = (Final_profit / Total_revenue) * 100
## Business_status = Profit if final profit > 0; loss if < 0; break-even if = 0


print("Total revenue : ",Total_revenue)
print("Total product cost : ",Total_product_cost)
print("Profit before advertising: ",Profit_before_advertising)
print("Final_profit: ",Final_profit)
print("Profit_margin: ",Profit_margin)
if (Profit_margin > 0):
    print("There is profit of ", Profit_margin)
elif (Profit_margin <0):
    print("There is Loss of ", Profit_margin)
else :
    print("There is neither profit nor Loss : ", Profit_margin)
```
